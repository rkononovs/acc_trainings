# Offline vision reasoning over an RGB grid

A complete, deterministic course companion.

## Learning objectives

- Represent a tiny image as in-memory RGB pixels.
- Calculate mean channel statistics.
- Classify the dominant colour deterministically without a download.

## Setup

The image is a 2-by-3 Python grid. No image library, remote asset, model, or credential is used by the offline cells.


For the optional live cell, install the dependency with `python -m pip install smolagents`.

## Complete implementation

Run the next cell to build the offline demonstration.

```python
rgb_grid = [
    [(20, 40, 220), (30, 50, 230), (10, 30, 210)],
    [(40, 60, 240), (25, 45, 225), (35, 55, 235)],
]

pixels = [pixel for row in rgb_grid for pixel in row]
channel_names = ("red", "green", "blue")
channel_means = {
    name: sum(pixel[index] for pixel in pixels) / len(pixels)
    for index, name in enumerate(channel_names)
}
dominant_colour = max(channel_means, key=channel_means.get)
vision_observation = {
    "width": len(rgb_grid[0]),
    "height": len(rgb_grid),
    "channel_means": channel_means,
    "dominant_colour": dominant_colour,
}
```

## Deterministic assertions

These checks make the lesson reproducible without a model or network connection.

```python
assert vision_observation["width"] == 3
assert vision_observation["height"] == 2
assert channel_means["blue"] > channel_means["green"] > channel_means["red"]
assert dominant_colour == "blue"

print("Offline assertions passed.")
```

## Results interpretation

Every pixel has a much larger blue value, and the aggregate statistics preserve that signal. Real vision agents may send images to a multimodal model or call specialized tools, but deterministic preprocessing remains valuable for tests and sanity checks.

## Credentials

Offline analysis needs none. The optional `smolagents` cell checks `HF_TOKEN`; an OpenAI-backed variant would separately check `OPENAI_API_KEY`. Neither value should be displayed or stored in the notebook.

```python
import os

live_model_provider = os.getenv("LIVE_MODEL_PROVIDER", "huggingface").strip().casefold()
hf_token = os.getenv("HF_TOKEN")
if live_model_provider != "local" and not hf_token:
    print("SKIPPED: Set HF_TOKEN to run this optional smolagents example.")
else:
    try:
        from smolagents import CodeAgent
        if live_model_provider == "local":
            from smolagents import TransformersModel
        else:
            from smolagents import InferenceClientModel
    except ImportError:
        print("SKIPPED: smolagents is not installed; run `python -m pip install smolagents`.")
    else:
        if live_model_provider == "local":
            model = TransformersModel(
                model_id="Qwen/Qwen2.5-Coder-1.5B-Instruct",
                device_map="cpu",
                torch_dtype="auto",
                max_new_tokens=256,
            )
        else:
            model = InferenceClientModel(
                model_id="Qwen/Qwen2.5-Coder-32B-Instruct", token=hf_token
            )
        openai_key_configured = bool(os.getenv("OPENAI_API_KEY"))
        live_agent = CodeAgent(tools=[], model=model, max_steps=2)
        live_result = live_agent.run(
            "Explain how mean RGB channels can provide a deterministic image sanity check."
        )
        print(live_result)
        if not openai_key_configured:
            print("SKIPPED: OPENAI_API_KEY is absent; no OpenAI-backed vision variant was attempted.")
```
